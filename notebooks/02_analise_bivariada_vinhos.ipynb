# Análise Bivariada - Vinhos Brasileiros

## Objetivo
Explorar relações entre pares de variáveis nos dados de vinhos, identificando correlações, padrões e associações.

## Estrutura da Análise
1. **Análises de Correlação**
2. **Gráficos de Dispersão** 
3. **Regressão Linear Simples**
4. **Testes de Hipóteses Bivariados**
5. **Análise de Associação**
6. **Comparação entre Grupos**
